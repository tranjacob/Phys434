### Importing python libraries and setting figure sizes

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

plt.rcParams["figure.figsize"] = (15,10)
```

### Drawing a sample of 1000000 values from a Gaussian Distribution centered around 67.0 with a standard deviation of 0.1

### The probability density function is given by $$f(x) = \frac{exp(-x^2/2)}{\sqrt{2\pi }}$$

```python
distribution = stats.norm.rvs(loc = 67.0, scale = .1, size = 1000000)
```

### Creating a plot of the distribution

* fig, ax creates the plot with 1 x-axis and 1 y-axis  
* ax.hist creates the normalized histogram with 50 bins using the sample named "distribution" from above  
* plt.tick_params sets the font size  
* plt.xlim sets the boundaries of x on the graph  
* np.linspace creates 1000 evenly spaced values between 66.5 and 67.5 for the function to be evaluated  
* ax.plot plots the histogram using the linspace values for x and the Gaussian distribution

```python
fig, ax = plt.subplots(1, 1)
ax.hist(distribution, 50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([66.5, 67.5])
x = np.linspace(66.5, 67.5, 1000)
ax.plot(x,stats.norm.pdf(x,loc = 67.0, scale = 0.1))
plt.show()
```
